Mathematical Operations in Series
NaN => Not a Number =>Missing Values

```python
import pandas as pd

series_1 = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
series_2 = pd.Series([10, 20, 30, 40, 50], index=['z', 'y', 'a', 'c', 'e'])

print(series_1)
```

```python
print(series_2)
```

```python
result_addition = series_1 + series_2
print(result_addition)
```

```python
import numpy as np
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([10,20,30,40,50])
print(arr1 + arr2)
```

```python
result = result_addition.fillna(0)
print(result)
```

```python
result_addition_dropped = result_addition.dropna()
print(result_addition_dropped)
```

Multiplication of two Series

```python
result_multiplication = series_1 * series_2
print(result_multiplication)
```

Division of two Series

```python
result_division = series_1 / series_2
print(result_division)
```
