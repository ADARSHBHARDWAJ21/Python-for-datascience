1) For Loop


```python
fruits =['apple', 'banana', 'cherry']
for fruit in fruits:
  print(fruit)
```

```python
nums = [2,6,4,5,7,9,10,11,12,15,17]
# write a condition to check which no are even and which are odd
for num in nums:
 if num%2 == 0:
  print(f'{num} is even')
 else:
  print(f'{num} is odd')
```

```python
input_str = 'AAABBBBCCCDDDDEEEE'
unique_str = ''
#write a code to remove duplicates
for char in input_str:
  if char not in unique_str:
    unique_str += char
print(unique_str)
#
```

List Comprehension- Combine the loop and an optional conditional expression into a single expression and return result in form of list

```python
squares = [x ** 2 for x in range(1,11)]
print(squares)
```

```python
even_numbers = [x for x in range(1,11) if x % 2 == 0]
print(even_numbers)
```

Nested for loop

```python
for i in range (1,6):
  for j in range(1,11):
    print(i*j, end='\t')
  print()
```
