12. Reverse a Two-Digit Number

Take a two-digit number and reverse it.

```python
num1 = int(input("Enter a 2 digith num: "))
# 1. Get the tens digit (e.g., for 47, this gives 4)
tens = num1 // 10
# 2. Get the ones digit (e.g., for 47, this gives 7)
ones = num1 % 10
reversed_num = (ones * 10) + tens

print("Reversed number:", reversed_num)
```

17. Time Conversion

Take a number of seconds and convert it into:

Hours
Minutes
Seconds

```python
num = int(input("Enter seconds: "))

# 1. Find total minutes and the leftover seconds
total_min = num // 60
rem_seconds = num % 60

# 2. Find total hours and the leftover minutes
hrs = total_min // 60
rem_min = total_min % 60

print("Hours:", hrs)
print("Minutes:", rem_min)
print("Seconds:", rem_seconds)
```
