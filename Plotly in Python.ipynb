```python
import plotly.express as px
data = px.data.iris()
data
```

```python
data['species'].value_counts()
```

Identify the relationship between sepal_width & sepal_length[two bumeric / continuous variables]
 # scatter plot

```python
fig = px.scatter(data, x='sepal_width', y='sepal_length', color='species')
fig.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.scatterplot(x= 'sepal_width', y= 'sepal_length', data= data, hue= 'species')
plt.xlabel('sepal_width')
plt.ylabel('sepal_length')
plt.show()
```

Line plot

Identify the trend of petal_length

```python
fig = px.line(data, y='petal_length', title = 'Petal Length Trend')

fig.show()
```

```python
fig = px.bar(data, x='species', y='sepal_length', title = 'sepal Length by species')

fig.show()
```
