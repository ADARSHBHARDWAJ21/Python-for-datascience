Logical Operators ->  are used to combine multiple conditionals statements or expressions

```python
x = 5
y =10
z = 15
print(x<y and y<z)#and ->true
print (x<y and y>z)#and -> false
print(x<y or y>z) # or -true
print(not(x<y)) # not -> false
```
