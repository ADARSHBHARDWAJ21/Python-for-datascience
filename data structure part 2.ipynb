Join set - 2 or more sets into a single set Union , intersection , Difference and so on

Union (use to combine 2 sets together) Operations

````python
set1 = { 1,2,3}
set2 = {3,4,5}

union_set = set1 | set2
print(union_set)
````

intersection operation

````python
intersection_set = set1 & set2
print(intersection_set)
````

difference operation

````python
difference_set = set1 - set2 #it will print the elements present in set 1 not in set 2
print(difference_set)
````

Symmetric difference
contains those elements that are present in either set 1 or set 2 but not in both

````python
symmetric_difference_set = set1 ^ set2
print(symmetric_difference_set)
````

Update() methods - updated methods modifies the original set itself

````python
set1.intersection_update(set2)
print(set1)
````

issubset()

````python
set1 = {1,2,3}
set2 = {1,2,3,4,5}
result = set1.issubset(set2)
print(result)
````

````python
result = set2.issuperset(set1)
print(result)
````
